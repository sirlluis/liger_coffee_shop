```python
import pandas as pd
from pathlib import Path
```

```python
DATA_PATH=Path().cwd()/"data"/"raw"
print(DATA_PATH)
```

```python
tablas = {
    "branches": None,
    "ingredients": None,
    "inventory_movements": None,
    "order_items": None,
    "orders": None,
    "products": None,
    "purchases": None,
    "recipes": None
}

for item in tablas:
    df = pd.read_csv(f"{DATA_PATH}/{item}.csv")
    column_names = df.columns.tolist()
    tablas[item] = column_names
```
